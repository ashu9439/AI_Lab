```python
# prompt: simple embedding example using bert with proper inline explanations

# Install necessary libraries
!pip install transformers sentence-transformers

# Import necessary libraries
from sentence_transformers import SentenceTransformer, util

# Load the pre-trained BERT model for sentence embeddings
model = SentenceTransformer('all-mpnet-base-v2') # Using all-mpnet-base-v2, you can change it to other models

# Example sentences
sentences = [
    "This is an example sentence.",
    "Each sentence is converted into an embedding vector.",
    "These vectors can be used for semantic similarity tasks."
]

# Compute sentence embeddings
embeddings = model.encode(sentences)

# Print the shape of the embeddings (number of sentences, embedding dimension)
print("Embeddings shape:", embeddings.shape)

# Calculate cosine similarity between sentence embeddings
cosine_scores = util.cos_sim(embeddings[0], embeddings[1]) # Comparing sentence 1 and sentence 2

# Print the cosine similarity score
print("Cosine similarity between sentence 1 and sentence 2:", cosine_scores[0][0])

# Example of finding semantically similar sentences:
query_embedding = model.encode("This is a test sentence related to the first example.")
similarity_scores = util.cos_sim(query_embedding, embeddings)

# Find the index of the most similar sentence
most_similar_index = similarity_scores.argmax()

# Print the most similar sentence and its score
print("Most similar sentence:", sentences[most_similar_index])
print("Similarity score:", similarity_scores[0][most_similar_index])
```
